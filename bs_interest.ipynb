### Problem setting

We consider a risky asset whose price foollows the stochastic process $S_t$, determined by the Black-Scholes SDE 
\begin{equation}
    dS_t = S_t(\mu dt + \sigma dW_t)
\end{equation}

With $(W_t)_t$ the standard brownian motion under the physical measure $\mathbb{P}$ 
We want to cover a european call option of payoff $F(t, S_T) = h(S_T)$ with a portfolio $\Pi = V - \Delta S_t$
(i.e We buy an option and sell the quantity $\Delta$ of the underlying), we consider $r_l$,$r_b$ respectively the lending and borrowing rate, meaning if $\Pi > 0$ we lend our money and grow it at the rate $r_l$ and if $\Pi \leq 0$ we borrow at the rate $r_b$ (the debt grows at the rate $r_b$)

Let $V_t$ be a self-financed portfolio, by definition $dV_t = H_{t}^{0} dS_{t}^{0} + H_t dS_t$.
By simplifying we get: $$dV_t = (H_{t}^{0}S_{t}^{0}r + H_t S_t \mu)dt + H_t S_t \sigma dW_t$$

considering $V$ a function of time $t$ and price $S$ we can write:
\begin{equation}
    \begin{split}
        dV &= V_t dt + V_S dS_t + \frac{1}{2} V_{SS} d \langle S , S \rangle \\
        &= (V_t + \frac{1}{2}\sigma^{2}S^{2}V_{SS} + \mu S V_S) dt + \sigma S V_S dW_t
    \end{split}
\end{equation}

if we chose $\Delta = V_S$, 
\begin{equation}
    \begin{split}
        d\Pi &= dV - V_S dS_t \\
        &= dV - V_S S (\mu dt + \sigma dW_t) \\
        &= (V_t + \frac{1}{2} \sigma^{2} S^{2} V_{SS})dt
    \end{split}
\end{equation}

We distinguish cases:
- if $\Pi > 0$, our portfolio being riskless (we removed the brownian term in the portfolio dynamic) by the absence of arbitrage opportunities in the BS model assumptions, we earn the lending rate $r_l$, meaning $d\Pi = r_l \Pi dt$ 

We then plug in both equations we have for $d\Pi$ and get :
\begin{equation}
    \begin{split}
        r_l \Pi &= V_t + \frac{1}{2} \sigma^2 S^2 V_{SS} \\
        r_l(V - S V_s) &= V_t + \frac{1}{2} \sigma^2 S^2 V_{SS} \\
        V_t + \frac{1}{2} \sigma^2 S^2 V_{SS} + r_l S V_s - r_l V &= 0
    \end{split}
\end{equation}

- if $\Pi \leq 0$, we pay interest at the $r_b$ rate, giving:
\begin{equation}
    V_t + \frac{1}{2} \sigma^2 S^2 V_{SS} + r_b S V_s - r_b V = 0
\end{equation}

Now we define the function $r$ as follows:
\begin{equation}
    r(c) =
    \begin{cases}
      r_l & \text{if}\ c > 0 \\
      r_b        & \text{otherwise}
    \end{cases}
\end{equation}

Then,
\begin{equation}
    V_t + \frac{1}{2} \sigma^2 S^2 V_{SS} - r(\Pi)\Pi = 0 \\
\end{equation}

as we have $-r(\Pi)\Pi = -r_l \Pi_+ + r_b \Pi_- = -r_l\Pi + (r_b - r_l) \Pi_-$
we replace and get the final expression of the PDE:
\begin{equation}
    V_t + \frac{1}{2} \sigma^2 S^2 V_{SS} + r_l S V_S + (r_b - r_l)(V - S V_S)_- - r_l V = 0
\end{equation}

We can now finally identify the terms of the general form semi-linear parabolic PDE and have:
(to simplify notation we set $\left[\sigma(t,x)\right]^{*}\nabla_x u = z$)

\begin{split}
    u &= V \\
    \mu(t,x) &= r_l x \\
    \sigma(t,x) &= \sigma x \\
    f(t,x,u, z) &= (r_b - r_l)(u - z/\sigma)_- - r_l u

\end{split}

We can now rewrite the problem as a BSDE, as there is an up-to-equivalence unique stochastic process $(X_t,Y_t,Z_t)_{t\in[0,T]}$ which satisfies:
\begin{equation}
\begin{split}
    X_t &= \zeta + \int_0^t r_l X_s ds + \int_0^t \sigma X_s dW_s \\
    Y_t &= h(X_T) + \int_t^T (r_b - r_l)(Y_s - Z_s/\sigma)_- - r_l Y_s ds - \int_t^T (Z_s)^{*} dW_s
\end{split}
\end{equation}

and u is a solution of the PDE iff $\qquad Y_t = u(t,X_t) \qquad Z_t = \left[\sigma(t,x)\right]^{*}\nabla_x u(t,X_t)$

### dimention 1 
given the above setup we can show that Feynmann-Kac cannot be used here, therefore we cannot simulate forward paths and average the payoffs to get the option price, only grid methods on the derived PDE and the neural network method on the BSDE survive this type of problem.

A standard Feynmann-Kac linear representation of the problem doesn't exist for this PDE.  

#### proof:
Suppose Feynmann-Kac is applicable and $r_l > r_b$, for the standard black-Scholes PDE, there exists a measure $\mathcal{Q}$ the price is then given by:

\begin{equation} 
    V(t,x) = E^{\mathcal{Q}} \left[ \exp^{-r(T-t)} \psi(X_T) \mid \mathcal{F}_t \right] 
\end{equation}

Now define, $H(g) \coloneqq E^{\mathcal{Q}} \left[ \exp^{-r(T-t)} g(X_T) \mid \mathcal{F}_t \right] $

Now, if $g=1$, to replicate the terminal payoff (i.e $1$) we lend at rate $r_l$, and then, $H(1) = E^{\mathcal{Q}} \left[ \exp^{-r_l(T-t)} \mid \mathcal{F}_t \right]$

if $g=-1$, we borrow at rate $r_b$ and $H(-1) = - E^{\mathcal{Q}} \left[ \exp^{-r_b(T-t)} \mid \mathcal{F}_t \right]$
but also, by linearity of the expectancy,

\begin{equation}
    \begin{split}
        H(-1) &= -H(1) \\
        - E^{\mathcal{Q}} \left[ \exp^{-r_b(T-t)} \mid \mathcal{F}_t \right] &= - E^{\mathcal{Q}} \left[ \exp^{-r_l(T-t)} \mid \mathcal{F}_t \right] \\
        r_b &= r_l 
    \end{split}
\end{equation}

Contradiction, as we assumed $r_l > r_b$, QED.

We use a A Rannacher-smoothed, $\theta$-scheme (Crank–Nicolson) finite-difference scheme with Picard iteration for the nonlinear reaction term, solved on a log-price grid ($x=logS$) via tridiagonal system solves: the method discretizes the linear diffusion–convection–reaction operator implicitly ($\theta$=0.5, falling back to fully implicit $\theta$=1.0 for the first rannacher_steps steps to damp the terminal-condition kink), linearizes the nonsmooth reaction term $(r_b−r_l)(V−SV_S)_{-}$ at each time step by fixed-point (Picard) iteration until the interior solution stabilizes below tolerance, and marches backward in time from the terminal payoff to recover $V(0,S_0)$ by interpolation on the grid.

In the particular case where $r_l = r_b$ we end up on the classic Black-Scholes European option price formula, a first good sanity check is to verify, the neural network and the finite differences method agree on these cases where we have an analytical solution available.

```python
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from deep_bsde_model import DeepBSDE
from bergman_1d import bergman_fd_1d
from sampling import call_payoff, eu_call_price
from tests import tests_1d
from functools import partial
import time

r_l = 0.01  # lending rate
r_b = 0.05  # borrowing rate
sigma = 0.2  # volatility
T = 1.0  # maturity<
N = 30 #standard time discretization
```

```python
r = 0.05
f_fn = lambda t,x,u,z: -r*u

K = np.linspace(80, 120, 5)
for K_val in K:
    price = eu_call_price(St=100, K=K_val, time_to_maturity=T, r=r, sigma=sigma)
    print(f"European call option price for K={K_val}: {price:.4f}")

    price = bergman_fd_1d(
        S0=100,
        K=K_val,
        T=T,
        r_l=r,
        r_b=r,
        sigma=sigma,
    )[0]
    print(f"Finite difference price for K={K_val}: {price:.4f}")
```

```python
from functools import partial


d = {}
K = np.linspace(80, 120, 5)

for K_val in K:
    payoff = partial(call_payoff, K=K_val)
    price_1 = eu_call_price(St=100, K=K_val, time_to_maturity=T, r=r, sigma=sigma)
    print(f"European call option price for K={K_val}: {price_1:.4f}")



    model = DeepBSDE(d=1, N=N, X0=100, y0=10.0, z0=np.ones(1), payoff_fn=payoff, mu_fn=lambda t,x: r*x, sigma_fn=lambda t,x: sigma*x, f_fn=f_fn, T=T, verbose=0)
    model.train(num_epochs=3000, batch_size=512)
    price_2 = model.y0.numpy()
    print(f"Neural network price for K={K_val}: {price_2:.4f}")

    d[K_val] = (price_1, price_2)
```

Now the interesting case is studying with $r_l \neq r_b$ if the NN agrees with the finite difference engine, we establish a list of test cases of different options whom we evaluate the price, every test case is stocked in disctionary inside a python file. 

```python
fd_output, nn_output = [], []

for name, test in tests_1d.items():
    print(f'testing the case: {name}')
    S0, K, T, r_l, r_b, sigma = test["S0"], test["K"], test["T"], test["r_l"], test["r_b"], test["sigma"]
    r_l, r_b, sigma = test["r_l"], test["r_b"], test["sigma"]
    f_fn = lambda t,x,u,z: (r_b - r_l) * tf.nn.relu(u - z/sigma) - r_l * u

    t0 = time.time()

    price_fd, delta, _ = bergman_fd_1d(S0=S0, K=K, T=T, r_l=r_l, r_b=r_b, sigma=sigma, payoff_fn=test["payoff"])
    elapsed = time.time() - t0
    print(f"Finite difference price: {price_fd:.4f}, time taken: {elapsed:.4f} seconds")
    fd_output.append((name, price_fd, delta, elapsed))

    t0 = time.time()

    #cheap initialization of y0 using the European call price
    y0_init = eu_call_price(St=S0, K=K, time_to_maturity=T, r=r_l, sigma=sigma)
    y0_init = float(np.ravel(y0_init)[0])

    nn_prices = []
    nn_deltas = []
    t0 = time.time()
    for i in range(3):
        tf.random.set_seed(i)

        t1 = time.time()
        model = DeepBSDE(num_units=100, d=1, N=N, X0=S0, y0=y0_init, z0=np.ones(1), payoff_fn=test["payoff"], 
                        mu_fn=lambda t,x: r_l*x, sigma_fn=lambda t,x: sigma*x, f_fn=f_fn, T=T, verbose=2)
        model.train(num_epochs=3000, batch_size=512)
        price_nn = model.y0.numpy()
        delta_nn = model.z0.numpy() / (sigma*S0)
        print(f"Neural network {i} price: {price_nn:.4f}, time taken: {time.time() - t1:.4f} seconds")
        nn_prices.append(price_nn)
        nn_deltas.append(delta_nn)

    elapsed = time.time() - t0
    print(f"Neural network average price: {np.mean(nn_prices):.4f}, standard deviation: {np.std(nn_prices):.4f}, time taken: {elapsed:.4f} seconds")
    nn_output.append((name, np.mean(nn_prices), np.mean(nn_deltas), elapsed))
```

```python
fd_df = pd.DataFrame(fd_output, columns=["Test Case", "FD Price", "FD Delta", "FD Time"])
nn_df = pd.DataFrame(nn_output, columns=["Test Case", "NN Price", "NN Delta", "NN Time"])

results = pd.merge(fd_df, nn_df, on="Test Case")
results = results[["Test Case", "FD Price", "NN Price", "FD Delta", "NN Delta", "FD Time", "NN Time"]]
results.to_csv("1d_bs_interest_results.csv", index=False)
```
